```python
class DenoisingModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.relu = nn.ReLU()
        self.layer1 = nn.Linear(16000, 1024 )
        self.layer2 = nn.Linear(1024, 1024 )
        self.layer3 = nn.Linear(1024, 16000 )
    def forward(self, x):
        x = self.layer1(x)
        x = self.relu(x)
        x = self.layer2(x)
        x = self.relu(x)
        x = self.layer3(x)
        return x
```

```python
model = DenoisingModel()
print(model)
```

```python
loss_fn = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001 )
```
